```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('housing.csv')
```

```python
df.head()
```

```python
##Q1
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='Landsize', y='Price', hue='Type')

plt.title('Property Price vs. Land Size (Colored By Property Type)')
plt.xlabel('Land Size')
plt.ylabel('Price')

plt.show()
```

```python
##Q3
yearly_avg_price = df.groupby('YearBuilt')['Price'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=yearly_avg_price, x='YearBuilt', y='Price')

plt.title('Average Property Price by Year Built')
plt.xlabel('Year Built')
plt.ylabel('Average Price (AUD)')

plt.show()
```

```python
##Q5
region_avg_price = df.groupby('Regionname')['Price'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(data=region_avg_price, x='Regionname', y='Price')

plt.title('Average Property Price by Region')
plt.xlabel('Region')
plt.ylabel('Average Price (AUD)')

plt.xticks(rotation=45)

plt.show()
```

```python
##Q7
plt.figure(figsize=(10, 6))

sns.histplot(df['Price'], bins=50, kde=True, color='teal')

plt.title('Distribution of Property Prices')
plt.xlabel('Price (AUD)')
plt.ylabel('Frequency')

plt.show()
```

```python
##Q9
plt.figure(figsize=(12, 8))

sns.boxplot(data=df, x='YearBuilt', y='Regionname')

plt.title('Distribution of Year Built by Region')
plt.xlabel('Year Built')
plt.ylabel('Region')

plt.show()
```
