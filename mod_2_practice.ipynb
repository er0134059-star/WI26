```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv('companies.csv')
```

```python
print(df.head())
```

```python
##Q1
plt.figure(figsize=(15,6))

sns.countplot(data=df, x='state')

plt.title('Count of Companies By State')
plt.xlabel('State')
plt.ylabel('Number of Companies')

plt.xticks(rotation=45)

```

```python
##Q3
plt.figure(figsize=(15, 8))

sns.barplot(data=df, x='Industry', y='job_openings', estimator=sum, errorbar=None)

plt.title('Total Job Openings By Industry')
plt.xlabel('Industry')
plt.ylabel('Total Job Openings')

plt.xticks(rotation=90)

plt.show()
```

```python
##Q5
plt.figure(figsize=(12, 6))

sns.lineplot(data=df, x='founded', y='growth_percentage', errorbar=None)

plt.title('Average Growth Rate By Year Founded')
plt.xlabel('Year Founded')
plt.ylabel('Average Growth Percentage')

plt.show()
```

```python
##Q7
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='employees', y='estimated_revenues')

plt.title('Employees vs. Estimated Revenues')
plt.xlabel('Number of Employees')
plt.ylabel('Estimated Revenues')

plt.show()
```

```python
##Q9
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x='founded', bins=30)

plt.title('Distribution of Companies By Year Founded')
plt.xlabel('Year Founded')
plt.ylabel('Number of Companies')

plt.show()
```
